```python

import pandas as pd

# Load datasets
matches = pd.read_csv("matches[1].csv", on_bad_lines="skip")
deliveries = pd.read_csv("deliveries[1].csv", on_bad_lines="skip")

# 1. Take number of rows from user
n = int(input("How many matches do you want to see? "))
print("\nHere are the", n, "matches:")
print(matches.head(n))

# 2. Take team name from user
team = input("\nEnter team name: ")
print("\nMatches played by", team)
team_data = matches[
    matches["team1"].str.contains(team, case=False, na=False)
    | matches["team2"].str.contains(team, case=False, na=False)
]
print(team_data[["team1", "team2", "winner"]])

# 3. Take player name from user (from deliveries dataset)
player = input("\nEnter batter name: ")
batter_col = "batter" if "batter" in deliveries.columns else "batsman"
runs_col = "batsman_runs" if "batsman_runs" in deliveries.columns else "total_runs"

player_data = deliveries[deliveries[batter_col].str.contains(player, case=False, na=False)]
print("\nTotal runs scored by", player, ":", player_data[runs_col].sum())
print(player_data[[batter_col, "bowler", runs_col]].head(10))
```
